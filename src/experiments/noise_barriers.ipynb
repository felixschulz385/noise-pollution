# Noise Barriers

Use the CLI directly.

List files in your Lastkajen account:

```bash
python src/cli.py data noise-barriers list-files
```

Fetch and preprocess by dataset name:

```bash
python src/cli.py data noise-barriers fetch --source highway --dataset-name Sound_Barriers_-_Road
python src/cli.py data noise-barriers preprocess --dataset-name Sound_Barriers_-_Road

python src/cli.py data noise-barriers fetch --source railway --dataset-name Sound_Barriers_-_Rail
python src/cli.py data noise-barriers preprocess --dataset-name Sound_Barriers_-_Rail
```
